# EEE-HPT STEP Viewer\nInteractively view the smoothed turbine blade STEP files.

```python
from pathlib import Path
import cadquery as cq
from ocp_vscode import show

STEP_DIR = Path(".").resolve() / "step"

STEP_FILES = {
    "hub_shroud": "hub_shroud.step",
    "stator1":    "stator1_smoothed.step",
    "rotor1":     "rotor1_smoothed.step",
    "stator2":    "stator2_smoothed.step",
    "rotor2":     "rotor2_smoothed.step",
}
```

## View a single part

```python
# View a single part
result = cq.importers.importStep(str(STEP_DIR / "stator1_smoothed.step"))
show(result)
```

## View all parts as an assembly

```python
# View all parts as a colour-coded assembly
colors = {
    "hub_shroud": "blue",
    "stator1":    "green",
    "rotor1":     "red",
    "stator2":    "purple",
    "rotor2":     "orange",
}

assembly = cq.Assembly(name="EEE-HPT")
for name, filename in STEP_FILES.items():
    full_path = STEP_DIR / filename
    if not full_path.exists():
        print(f"  Skipping {name}: not found")
        continue
    print(f"  Loading {name} ...")
    shape = cq.importers.importStep(str(full_path))
    assembly.add(shape, name=name, color=cq.Color(colors.get(name, "gray")))

show(assembly)
```
